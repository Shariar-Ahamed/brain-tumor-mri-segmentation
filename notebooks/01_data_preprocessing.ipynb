```python
# Preprocessing Utility Functions
def zscore_norm(vol):
    mask = vol > 0
    if not np.any(mask):
        return vol
    mean, std = np.mean(vol[mask]), np.std(vol[mask])
    norm = np.zeros_like(vol)
    norm[mask] = (vol[mask] - mean) / (std + 1e-8)
    return norm

def remap_labels(mask):
    m = np.zeros_like(mask)
    m[mask == 1] = 1 # NCR/NET
    m[mask == 2] = 2 # ED
    m[mask == 4] = 3 # ET
    return m

print("Z-score & Label Remapping ready.")
```
